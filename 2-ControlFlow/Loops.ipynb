```python
## Strings

str="Henry Prasetya"

for i in str:
    print(i)
```

```python
## while loop

## it continues to execte as long as condition is true

count=0

while count<10:
    print(count)
    count=count+1
```

```python
## Loop Control Statements

## break
## the break statement exits the loop prematurly

## break statements

for i in range(10):
    if i==5:
        break
    print(i)
```

```python
# continue

## skip iteration and continue the next

for i in range(10):
    if i%2==0:
        continue
    print(i)
```

```python
## pass
## pass statement is null operation, it does nothing.

for i in range(5):
    if i==3:
        print("the number is",i)
        pass
    print(i)
```

```python
## nested loops
## a loop inside a loop

for i in range(3):
    for j in range(2):
        print(f"i:{i} and j:{j}")
```

```python
## Examples- Calculate the sum of first N natural number using while and for loop

## while loop

n=10
sum=0
count=1

while count<=n:
    sum=sum+count
    count=count+1

print("sum of first 10 natural number:",sum)
```

```python
n=10
sum=0
count=1

for i in range(11):
    sum=sum+i

print(sum)
```

```python
## Example = prime numbers between 1 and 100

for num in range(1,101):
    if num>1:
        for i in range(2,num):
            if num%i == 0:
                break
        else:
            print(num)
```
