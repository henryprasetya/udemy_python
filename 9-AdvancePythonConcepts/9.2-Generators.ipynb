### Generator

#### Generator adalah cara simpel untuk membuat iterator dengan kondisi tidak menyimpannya di memory


```python
def square(n):
    for i in range(3):
        yield i**2      # yield merupakan kunci untuk menggunakan generator
```

```python
square(3)
```

```python
for i in square(3):
    print(i)
```

```python
a=square(3) # Akan diidentifikasi sebagai generator karena di baris 5 awal tadi sudah menggunakan yield
a
```

```python
next(a) # Iterasi pangkat 2 mulai dari 0 - 3 sampai error
```

```python
def my_generator():
    yield 1
    yield 2
    yield 3
```

```python
gen=my_generator()
gen
```

```python
next(gen)
```

```python
for val in gen:
    print(val)
```

### Practical Example : Reading Large Files
#### Generator sangat efisien digunakan untuk membaca file berukuran besar karena bisa hanya membaca 1 baris saja sesuai kebutuhan tanpa load semua file ke memory

```python
def read_large_file(file_path):
    with open(file_path, 'r') as file:
        for line in file:
            yield line
```

```python
file_path='large_file.txt'

for line in read_large_file(file_path):
    print(line.strip())
```
