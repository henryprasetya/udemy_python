#### Decorator di Python adalah fungsi khusus yang digunakan untuk memodifikasi perilaku method (fungsi dalam class) tanpa mengubah kodenya secara langsung.

```python
## Function Copy

def welcome():
    return "Selamat datang di python advanced course"

welcome()
```

```python
wel=welcome
print(wel())
del welcome
print(wel())
```

```python
## Closures Function
## Secara singkat Function di dalam function
# Contoh 1
def main_welcome():
    msg="Selamat Datang"
    mgs="Test"
    def sub_welcome_method():
        print("Selamat datang di python advanced course")
        print(msg)
        print(mgs)
        print("Tolong pelajari konsep ini dengan baik")
    return sub_welcome_method()
```

```python
main_welcome()
```

```python
# Contoh 2
def main_welcome(msg):

    def sub_welcome_method():
        print("Selamat datang di python advanced course")
        print(msg)
        print("Tolong pelajari konsep ini dengan baik")
    return sub_welcome_method()
```

```python
main_welcome("Selamat datang semuanya")
```

```python
# Contoh 3
def main_welcome(func):

    def sub_welcome_method():
        print("Selamat datang di python advanced course")
        func("Selamat datang di tutorial ini")
        print("Tolong pelajari konsep ini dengan baik")
    return sub_welcome_method()
```

```python
main_welcome(print)
```

```python
# Contoh 4
def main_welcome(func,lst):

    def sub_welcome_method():
        print("Selamat datang di python advanced course")
        print(func(lst))
        print("Tolong pelajari konsep ini dengan baik")
    return sub_welcome_method()
```

```python
### Decorator

def main_welcome(func):

    def sub_welcome_method():
        print("Selamat datang di python advanced course")
        func()
        print("Tolong pelajari konsep ini dengan baik")
    return sub_welcome_method()
```

```python
def course_introduction():
    print("Ini adalah advanced python code")

course_introduction()
```

```python
main_welcome(course_introduction)
```

```python
@main_welcome
def course_introduction():
    print("ini adalah advanced python course")
```

```python
## Decorator

def my_decorator(func):
    def wrapper():
        print("sebelum function dipanggil")
        func()
        print("sesudah function dipanggil")
    return wrapper
```

```python
@my_decorator
def say_hello():
    print("Halloo")
```

```python
say_hello()
```

```python
## Decorators with arguments

def repeat(n):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(n):
                func(*args, **kwargs)
        return wrapper
    return decorator
```

```python
@repeat(3)
def say_hello():
    print("Helloo")
```

```python
say_hello()
```
