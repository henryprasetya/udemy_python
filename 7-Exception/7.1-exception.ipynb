Exception berfungsi untuk menghandle error tanpa menghentekan eksekusi sebuah program yang dijalankan
Kurang lebih hampir mirip dengan if elif else

Jenis exceptions

1. ZeroDivisionError: Dividing by zero
2. FileNotFoundError: File not found
3. ValueError  : invalid value
4. TypeError: Invalid type

```python
## Exception try, except block

try:
    a=b
except:
    print("The variable has not been assigned")
```

```python
# Pesan error akan keluar namun tidak seperti di baris 1
try:
    a=b
except NameError as ex: #ex sebagai alias
    print(ex)
```

```python
## Zero Division Error
result=1/0
```

```python
try:
    result=1/0
except ZeroDivisionError as ex:
    print(ex)
    print("please enter the denominator greater than 0")
```

```python
## Pasti error karena handlenya hanya untuk ZeroDivisionError
try:
    result=1/2
    a=b
except ZeroDivisionError as ex:
    print(ex)
    print("please enter the denominator greater than 0")
```

```python
# Exception bisa untuk handle banyak kelas
try:
    result=1/2
    a=b
except ZeroDivisionError as ex:
    print(ex)
    print("please enter the denominator greater than 0")
except Exception as ex1:
    print(ex1)
    print('Main exception got caught here')
```

```python
# Except Block
try:
    num=int(input("Enter a number"))
    result=10/num
except ValueError:
    print("This is not a valid number")
except ZeroDivisionError:
    print("enter denominator greater than 0")
except Exception as ex:
    print(ex)
```

```python
## Try, Except, Else Block

try:
    num=int(input("Enter a number"))
    result=10/num
except ValueError:
    print("That's not a valid number!")
except ZeroDivisionError:
    print("You Cant devide by zero!")
except Exception as ex:
    print(ex)
else:
    print(f"the result is {result}")        # Kalau tidak mendapat exception maka baris kode ini yang akan dieksekusi
```

```python
## try, except, else, finally
try:
    num = int(input("Enter a number: "))
    result = 10 / num
except ValueError:
    print("That's not a valid number!")
except ZeroDivisionError:
    print("You can't devide by zero")
else:
    print(f"The result is {result}")        # Akan dieksekusi ketika tidak ada error atau bisa melewati block exception
finally:
    print("Execution complete")             # Fungsi finally akan dieksekusi setelah block exception ada error maupun tidak
```
