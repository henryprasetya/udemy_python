### Methods Overriding
subclass membuat ulang method yang ada pada superclass

```python
## Base class

class Animal:
    def speak(self):
        return "Suara dari animal"
    
## Derived class 1
# Methodnya sama dengan animal

class Dog(Animal):
    def speak(Self):
        return "Woof"
    
## Derived class 2

class Cat(Animal):
    def speak(self):
        return "Meaw"

## Function yang mendemonstrasikan polymorphism
def animal_speak(animal):
    print(animal.speak())

dog=Dog()
print(dog.speak())

cat=Cat()
print(cat.speak())

animal_speak(dog)
```

```python
### Polymorphism with functions and methods
## Base class

class Shape:
    def area(self):
        return "Luas bangun datar"
    
## Derived class

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width=width
        self.height=height

    def area(self):
        return self.width * self.height
    
## Derived class 2

class Circle(Shape):
    def __init__(self, radius):
        self.radius=radius

    def area(self):
        return 3.14 * self.radius * self.radius
    
## Function mendemonstrasikan polymorphism

def print_area(shape):
    print (f"Luasnya adalah {shape.area()}")

rectangle=Rectangle(4,5)
circle=Circle(3)
bentuk=Shape()

print_area(rectangle)
print_area(circle)
print_area(bentuk)
```

```python
## Polymorphism with abstract base classes

from abc import ABC, abstractmethod

## Define an abstract class

class Vehicle(ABC):
    @abstractmethod
    def start_engine(self):
        pass

## Derived class 1

class Car(Vehicle):
    def start_engine(self):
        return "Car engine started"

## Derived class 2

class Motorcycle(Vehicle):
    def start_engine(self):
        return "Motorcycle engine has been started"
    
## Polimorphism

def start_vehicle(vehicle):
    print(vehicle.start_engine())
    
## Create object of car and motorcycle

car = Car()
motorcycle = Motorcycle()

start_vehicle(car)
start_vehicle(motorcycle)
```
