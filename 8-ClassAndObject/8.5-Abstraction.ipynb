konsep OOP untuk menyembunyikan detail implementasi dan hanya menampilkan bagian penting (fungsi/fitur) agar lebih sederhana digunakan.

Sebagai contoh tombol shutdown di windows, dibalik itu semua terdapat proses dan user hanya mengetahui tombol shutdown dan loading screennya saja

```python
from abc import ABC,abstractmethod

## Abstract base class

class Vehicle(ABC):
    def drive(self):
        print("The vehicle is ready to use") # Ini metoda normal

    @abstractmethod
    def start_engine(self):
        pass

class Car(Vehicle):
    def start_engine(self):
        print("Car engine started")

def operate_vehicle(vehicle):
    vehicle.start_engine()
    vehicle.drive()

car=Car()
operate_vehicle(car)
```
