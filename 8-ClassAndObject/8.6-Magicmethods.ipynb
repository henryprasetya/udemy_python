#### Magic methods (kadang disebut juga dunder methods, dari double underscore) adalah method khusus di Python yang namanya diawali dan diakhiri dengan dua garis bawah. Magic methods membuat class dan object kamu bisa berperilaku seperti tipe data bawaan Python (seperti int, str, list, dll).

```python
class Person:
    pass

person=Person()
dir(person) # Cara mengetahui magic methods
```

```python
print(person) # pesan ini berasal dari fungsi __str__
```

```python
## Basic Magic Methods
class Person:
    def __init__(self, name, age):
        self.name=name
        self.age=age

person=Person("Henry", 55)
print(person) # Output berupa __str__ yang harus di overide bila ingin memunculkan value dari variable person
```

```python
## Basic Magic Methods
class Person:
    def __init__(self, name, age):
        self.name=name
        self.age=age

    def __str__(self):
        return f"{self.name},{self.age} years old"
    
    def __repr__(self):
        return f"Person=(name={self.name}, age={self.age})"
    
person=Person("Henry", 55)
print(person)
print(repr(person))
```
