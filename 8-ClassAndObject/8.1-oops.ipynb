```python
### Instance Variable and Methods
class Dog:
    ## constructor
    def __init__(self,name,age):
        self.name=name
        self.age=age

## Create Objects
dog1=Dog("Budi",3)
print(dog1)
print(dog1.name)
print(dog1.age)
```

```python
dog2=Dog("Lusi",4)
print(dog2.name)
print(dog2.age)
```

```python
## Define class with instance methods

class Dog:
    def __init__(self,name,age):
        self.name=name
        self.age=age

    def bark(self):
        print(f"{self.name} says wooooof")
        print(f"Umurnya {self.age} ")

dog1=Dog("budri",5)
dog1.bark()
dog2=Dog("Luci",6)
dog2.bark()
```

```python
### Modeling a bank Account

## Define class for bank account

class BankAccount:
    def __init__(self,owner,balance=0):
        self.owner=owner
        self.balance=balance
    
    def deposit(self,amount):
        self.balance+=amount
        print(f"{amount} telah di deposit, balance baru adalah {self.balance}")

    def withdraw(self,amount):
        if amount>self.balance:
            print("Saldo kurang")
        else:
            self.balance-=amount
            print(f"{amount} telah di withdraw, balance sekarang {self.balance}")

    def get_balance(self):
        return self.balance
    
##
account=BankAccount("Henry",10000)
print(account.balance)
```

```python
## Call instance methods

account.deposit(1000)
```

```python
account.withdraw(500)
```

```python
account.get_balance()
print(account.get_balance())    # Output sama saja
```
