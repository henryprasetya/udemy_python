Python menyediakan built-in function dan methods untuk read dan write files

```python
### Read a Whole File

with open('example.txt', 'r') as file:  #Buka File dengan mode read mode
    content=file.read()
    print(content)
```

```python
## Read a file line by line

with open('example.txt','r') as file:
    for line in file:
        print(line)
```

```python
## Read a file line by line

with open('example.txt','r') as file:
    for line in file:
        print(line.strip())             # .strip() remove the newline character
```

```python
## Writing a file (overwriting)

with open('example.txt','w') as file:       #'w' overwrite mode berfungsi untuk merubah, menambah, mengurangi isi dari sebuah file
    file.write('Hello World!\n')            #\n berfungsi untuk perintah new line/baris baru
    file.write('Ini adalah baris baru.')
```

```python
## Write a file without overwriting

with open('example.txt','a') as file:               # 'a' append mode berfungsi menambah isi tanpa mengurangi isi dari file
    file.write("Append operation taking place!\n")    
```

```python
## Writing a list of line to a file
lines=['First line \n','Second line \n','Third line \n']
with open('example.txt','a') as file:
    file.writelines(lines)
```

```python
### Binary Files 
##  biasanya file berekstensi .bin

# Writing to a binary file

data = b'\x00\x01\x02\x03\x04'
with open('example.bin','wb') as file:  # 'wb' adalah writebyte
    file.write(data)
```

```python
## Read binary file
with open('example.bin','rb') as file:  # 'rb' adalah readbyte
    content = file.read()
    print(content)
```

```python
### Read the content from a source text file and write to a destination text file

with open('example.txt','r') as source_file:
    content = source_file.read()

with open('destination.txt','w') as destination_file:
    destination_file.write(content)
```

```python
## Read a text file and count the number of lines, words, and characters

def count_text_file(file_path):
    with open(file_path,'r') as file:
        lines = file.readlines()
        line_count = len(lines)
        word_count = sum(len(line.split()) for line in lines)
        char_count = sum(len(line) for line in lines)
    return line_count, word_count, char_count

file_path = 'example.txt'
lines, words, characters = count_text_file(file_path)
print(f'lines: {lines} words: {words} characters: {characters}')
```

```python
### Writing and the reading a file

with open('example.txt','w+') as file:  # 'w+' berfungsi untuk membuka file untuk read dan write, kalau file tidak ada maka akan dibuat file baru.
    file.write("Hello World\n")
    file.write("This is a new line \n")

    ## Move the file cursor to the beginning
    file.seek(0)

    # Read the content of the file
    content=file.read()
    print(content)
```
