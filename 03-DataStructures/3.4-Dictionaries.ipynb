Dictionaries adalah sebuah koleksi item yang tidak terstruktur atau tidak runtut.

```python
## Creating Dictionaries

# [] berarti bikin list kosong
# () berarti bikin tuple kosong

empty_dict={}
print(type(empty_dict))
```

```python
empty_dict=dict()
empty_dict
```

```python
student={"name":"Henry","age":27,"grade":24}
print(student)
print(type(student))
```

```python
# Error example, key should be unique, single key always used
student={"name":"Henry","age":27,"name":24}
print(student)
```

```python
## Accessing Dictionary Elemenets

student={"name":"Henry","age":27,"grade":"A"}
print(student)
```

```python
## Accessing Dictionary Elements (salah satu)
print(student['grade'])
print(student['age'])

## Accessing using get() method
print(student.get('grade'))
print(student.get('last_name'))
print(student.get('last_name',"Not Available"))
```

```python
## Modifying Dictionary Element
## Dictionary are mutable, jadi bisa add, update, delete elements

print(student)
```

```python
student["age"]=33 #Update value dari age
print(student)
student["address"]="Indonesia" # Add new key dan value
print(student)
```

```python
## Delete Key

del student['grade'] # Delete Key and value
print(student)
```

```python
## Dictionary Methods

keys=student.keys() # Get all the keys
print(keys)
values=student.values() # Get all the values
print(values)

items=student.items() # Get all the key values pairs
print(items)
```

```python
## Shallow copy = Copy key/value

student_copy=student
print(student)
print(student_copy)
```

```python
student['name']="Henry2"
print(student)
print(student_copy)
```

```python
student_copy1=student.copy() #agar key/value utama tidak berubah
print(student_copy1)
print(student)
```

```python
student["name"]="Heeenry"
print(student_copy1)
print(student)
```

```python
## Iterating over Dictionaries
## bisa pakai loops untuk iterate over dictionaries (key, values, or items)

## Iterating over keys
for keys in student.keys():
    print(keys)
```

```python
## Iterate over values

for value in student.values():
    print(value)
```

```python
## Iterate over key value pairs

for key,value in student.items():
    print(f"{key}:{value}")
```

```python
## Nested Dictionaries

students={
    "student1":{"name":"Henry","age":"28"},
    "student2":{"name":"Vorne","age":"30"}
}
print(students)
```

```python
### Access Nested Dictionaries elements

print(students["student2"]["name"])
print(students["student2"]["age"])
```

```python
students.items()
```

```python
## Iterating over nested Dictionaries

for student_id,student_info in students.items():
    print(f"{student_id}:{student_info}")
    for key,value in student_info.items():
        print(f"{key}:{value}")
```

```python
## Dictionary Comprehension

squares={x:x**2 for x in range(5)}
print(squares)
```

```python
## Condition Dictionary comprehension

evens={x:x**2 for x in range(10) if x%2==0}
print(evens)
```

```python
## Practice Examples

## Use a Dictionary to count the frequency of elements in list

numbers=[1,2,2,3,3,3,4,4,4,4]
frequency={}

for number in numbers:
   if number in frequency:
      frequency[number] += 1
   else:
      frequency[number] = 1

print(frequency)

```

```python
## Merge 2 Dictionaries into 1

dict1 = {"a" : "1" , "b" : "2"}
dict2 = {"b" : "3" , "c" : "4"}
merged_dict = {**dict1, **dict2}
print(merged_dict)
```
