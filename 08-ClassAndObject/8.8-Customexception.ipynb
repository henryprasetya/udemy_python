### Custom Exception (Raise and Throw and exception)

```python
class Error(Exception): # Inherit Exception class
    pass

class dobException(Error): # Inherit Error Class, dob = Date of Birth
    pass

## Tujuannya nanti agar bisa customize pesan yang ditampilkan untuk hal tertentu
```

```python
year=int(input("Enter the dob"))
age=2024-year

try: # Try catch block
    if age<=30 and age>20:
        print("The age is valid, you can apply for exam")
    else:
        raise dobException # harus ditambah try catch block sebelum raise exception
    
except dobException:
    print("Maaf, umur tidak sesuai dengan kriteria, harus kurang dari 30 dan lebih dari 20")
```
