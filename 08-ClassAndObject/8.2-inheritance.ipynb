```python
## Inheritance (Single Inheritance)
## Class bisa mewarisi atau memakai methods yang ada pada class lain atau yang sudah ada
## Parent class

class Car:
    def __init__(self, windows, doors, enginetype):
        self.windows = windows
        self.doors = doors
        self.enginetype = enginetype
    
    def drive(self):
        print(f"Seseorang akan mengendarai mobil dengan jenis mesin {self.enginetype}")
```

```python
car1=Car(4,5,"Bensin")
car1.drive()
#Contoh lain
#print(car1.drive())
```

```python
# Inherite example
class Tesla(Car):   # Di dalam bracket terdapat (Car) bertujuan untuk mewarisi atau bisa memakai methods pada class Car
    def __init__(self, windows, doors, enginetype,is_selfdriving):
        super().__init__(windows, doors, enginetype)  # super berfungsi untuk memanggil parent class, disini sebagai contoh parent classnya dari __init__ class Car
        self.is_selfdriving = is_selfdriving

    def selfdriving(self):
        print(f"Apakah Tesla support dengan mengemudi otomatis : {self.is_selfdriving}")
```

```python
tesla1=Tesla(4,5,"Elektric",True)
tesla1.selfdriving()
```

```python
tesla1.drive()
```

```python
### Multiple Inheritance
## Menggabungkan 2 class yang sudah dibuat
## Base class

class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print("Subclass must implement this method")

## Base class 2
class Pet:
    def __init__(self, owner):
        self.owner = owner

## Derived class
# Inherite/menggabung 2 class tersebut

class Dog(Animal, Pet):
    def __init__(self, name, owner):
        Animal.__init__(self, name)
        Pet.__init__(self, owner)

    def speak(self):
        return f"{self.name} menggonggong"
    
## Create and object
dog=Dog("Budi","Henry")
print(dog.speak())
print(f"Owner:{dog.owner}")
```
