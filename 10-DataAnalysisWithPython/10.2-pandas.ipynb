### Pandas-Dataframe and Series

#### Panda merupakan library data manipulation di python, digunakan untuk data analysis dan data cleaning. Panda menyediakan 2 struktur data yaitu Series dan DataFrame. Series merupakan One Dimension Array-like object, Kalau DataFrame merupakan Two-dimensional, size-mutable.

```python
import pandas as pd
```

```python
## Series
## Panda Series merupakan 1 Dimensional Array-like object bisa hold berbagai jenis data. mirip column pada sebuah tabel

import pandas as pd
data = [1,2,3,4,5]
series=pd.Series(data)

print("Series \n", series)
print(type(series))
```

```python
## Create Series from Dictionary

data = {'a':1,'b':2,'c':3}
series_dict=pd.Series(data)
print(series_dict)
```

```python
data = [10, 20, 30]
index = ['a','b','c']
pd.Series(data, index=index)
```

```python
## DataFrame
## Two Dimensional, terdapat rows dan columns, mirip table pada database / excel

## Create a Dataframe from dictionary oof list, Cara pertama
data={
    'Name':['Henry','Jack','Milton'],
    'Age':[25,31,44],
    'City':['Malang','Tokyo','Texas']
}
df=pd.DataFrame(data)   # Create DataFrame berdasarkan variable data
print(df)
print(type(df))
```

```python
## Create DataFrame from list of dictionaries
## Cara kedua

data=[
    {'Name':'Henry','age':22,'City':'Malang'},  # First row
    {'Name':'Darn','age':22,'City':'Malang'},  # Second row
    {'Name':'Lenno','age':22,'City':'Malang'},
    {'Name':'Jack','age':22,'City':'Malang'}
]
df=pd.DataFrame(data)
print(df)
print(type(df))
```
