#### Iterators merupakan sebuah advanced concepts yang memiliki keunggulan untuk looping secara efisien pada memory manajemennya

```python
my_list=[1,2,3,4,5,6]
for i in my_list:
    print(i)
```

```python
type(my_list)
```

```python
## Iterator

iterator=iter(my_list)
print(type(iterator))
```

```python
iterator
```

```python
## Iterate throught all elements with next

next(iterator)  # akan bertambah secara satu per satu sampai valuenya habis
```

```python
iterator=iter(my_list)
```

```python
# Agar ketika di iterasi dengan next tidak error bila elemennya habis

try:
    print(next(iterator))
except StopIteration:
    print("Tidak ada element di iterator")
```

```python
## String iterator

my_string = "Haloo"
string_iterator = iter(my_string)

print(next(string_iterator))    # Output : H
print(next(string_iterator))    # Output : a
```
