1. Manage to create tot do list

Create to do list 

```python
to_do_list=["Beli belanjaan","Membersihkan rumah","Bayar tagihan"]

##Adding to task
to_do_list.append("Jadwal meeting")
to_do_list.append("Latihan lari")

##Removing a completed task

to_do_list.remove("Membersihkan rumah")

## Checking if a task is in a list
if "Bayar tagihan" in to_do_list:
    print("Jangan lupa bayar tagihan")

print("To Do List Remaining")
for task in to_do_list:
    print(f"-{task}")
```

Example 2: Organizing Student Grades
    Create lust to store and calculate avarage grades for students

```python
# Organizing students grades
grades = [85, 92, 78, 90, 88]

# Adding a new grade
grades.append(95)

# Calculating the average grade
average_grades = sum(grades) / len(grades)
print(f"Average Grade: {average_grades:.2f}")

# Finding the highest and lowest grades
highest_grades = max(grades)
lowest_grades = min(grades)
print(f"Highest Grades: {highest_grades}")
print(f"Lowest Grades: {lowest_grades}")
```

Example 3: Managing An Inventory

Use a list to manage inventory items in a store

```python
# Managing an inventory

inventory = ["apel", "pisang", "jeruk", "anggur"]

# Adding new item
inventory.append("strawberry")

# Removing an item
inventory.remove("pisang")

# Checking if and item is in stock
item = "jeruk"
if item in inventory:
    print(f"{item} stoknya ada")
else:
    ("{item} stoknya habis")

# Printing the inventory
print("Inventory List:")
for item in inventory:
    print(f"- {item}")
```

Example 4 : Collecting user feedback

Use a list to collect and ananylze user feedback

```python
# Collecting user feedback

feedback = ["Great service!", "Very satisfied", "Could be better", "Excellent experience"]

# Adding new feedback
feedback.append("Not happy with the service")

# Counting specific feedback
positive_feedback_count = sum(1 for comment in feedback if "great" in comment.lower() or "excellent" in comment.lower())
print(f"Positive Feedback Count: {positive_feedback_count}")

# Printing all feedback
print("User Feedback:")
for comment in feedback:
    print(f"- {comment}")
```
