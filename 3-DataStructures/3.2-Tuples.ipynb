```python
## Immutable Nature of Tuple
## Tuples are immutable, meaning their elements cannot be change once assigned

lst=[1,2,3,4,5]
print(lst)

lst[1]="Henry"
print(lst)

# So list are mmutable (bisa diubah ubah/otak atik)
```

```python
## Tuple Methods
print(numbers.count(1)) # hitung berapa kali nomor yang dicari muncul
print(numbers.index(3))
```

```python
## Packing and unpacking tuple
## Packing
packed_tuple=1, "Hello", 3.14
print(packed_tuple)
```

```python
## Unpacking a tuple

a,b,c=packed_tuple

print(a)
print(b)
print(c)
```

```python
## Unpacking with *

numbers=(1,2,3,4,5,6)
first,*middle,last=numbers
print(first)
print(middle)
print(last)
```

```python
## Nested Tuple
## Nested List
lst=[[1,2,3,4],[6,7,8,9],["Hello",3.14,"c"]]
lst[0][0:3] # disini coba panggil data dari index ke 0, di dalam isi index 0 yang awal isinya angka 1
```

```python
nested_tuple = ((1,2,3), ("a", "b", "c"),(True, False))

## Access the element inside a tuple
print(nested_tuple[0])
print(nested_tuple[1][2])
```

```python
## Iterating over nested tuple
for sub_tuple in nested_tuple:
    for item in sub_tuple:
        print(item,end=" ")
    print()
```
