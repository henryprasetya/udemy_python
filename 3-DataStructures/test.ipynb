```python
numbers=[1,2,2,3,3,3,4,4,4,4]
frequency={}

for number in numbers:
    if number in frequency:
        frequency[number] +=1
    else:
        frequency[number] =1

print(frequency)
```

```python
favorite_fruits=["Apel", "Jeruk", "Apel", "Mangga", "Jeruk", "Pisang", "Apel", "Mangga", "Pisang", "Jeruk"]
fruit_counts = {}

for fruit in favorite_fruits:
    if fruit in fruit_counts:
        fruit_counts[fruit] += 1
    else:
        fruit_counts[fruit] = 1

print(fruit_counts)
    
```
