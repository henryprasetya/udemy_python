```python
## Create a set
my_set={1,2,3,4,5}
print(my_set)
print(type(my_set))
```

```python
# Basic Sets Operation
# Adding and removing element

my_set.add(7)
print(my_set)
my_set.add(7) # it will ignore because it already exist
print(my_set)
```

```python
## Removing the elemet from set

my_set.remove(3)
print(my_set)
```

```python
my_set.remove(10)
```

```python
my_set.discard(11)
print(my_set)
```

```python
# Pop Method
remove_element = my_set.pop()
print(remove_element)  # Output: 3
print(my_set)
```

```python
## Clear all the element

my_set.clear()
print(my_set)
```

```python
## Set Membership test

my_set={1,2,3,4,5}
print(3 in my_set)
print(10 in my_set)
```

```python
## Mathematical Operation
set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}

## Union for combining element for example set1 and set2
union_set=set1.union(set2)
print(union_set)

## Intersection for showing common element for example set1 and set2
intersection_set=set1.intersection(set2)
print(intersection_set)

set1.intersection_update(set2)
print(set1)
```

```python
set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}

## Difference = mencari angka yang berbeda

print(set1.difference(set2))
```

```python
## Symmetric Difference = hanya cari yang unik satu sama lain, yang sama akan dihapus

set1.symmetric_difference(set2)
```

```python
## Sets methods

set1={1,2,3,4,5}
set2={3,4,5}

## is subset

print(set1.issubset(set2)) #cek apakah set1 berada di dalam set2

print(set1.issuperset(set2)) #cek apakah set 1 punya semua element di set2
```

```python
## Counting unique words in text

text="In this tutorial we are discussing about sets"
words=text.split()

## Convert list of words to set to get unique words

unique_words=set(words)
print(unique_words)
print(len(unique_words))
```
