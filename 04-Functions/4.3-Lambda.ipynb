```python
def addition(a,b):
    return a+b
```

```python
addition(2,3)
```

```python
addition=lambda a,b:a+b
type(addition)
print(addition(5,6))
```

```python
#without lambda
def even(num):
    if num%2==0:
        return True
    else:
        return False
even(23)
```

```python
# With Lambda which is more simple
even1=lambda num:num%2==0
even1(12)
```

```python
## Multiple parameters

def addition(x,y,z):
    return x+y+z

addition(12,13,14)
```

```python
addition1=lambda x,y,z:x+y+z
addition1(12,13,14)
```

```python
## map() - applies funcion to all item in the list
# untuk materi 4.4

numbers=[1,2,3,4,5,6]
def square(number):
    return number**2

square(4)
```

```python
list(map(lambda x:x**2,numbers))
```
