```python
## Default Parameters

def greet(name):
    print(f"Hello {name} Selamat datang")

greet("Henry")
```

```python
def greet(name="Tamu"):
    print(f"Hello {name} Selamat datang")

greet()
```

```python
greet("Henryy")
```

```python
## Variable Lenght Arguments
## Positional and Keywords Arguments
#pakai *args agar tidak tulis berulang
def print_numbers(*henry):
    for numbers in henry:
        print(numbers)
```

```python
print_numbers(1,2,3,4,5,6,7,8,"Henry")
```

```python
def print_numbers(*args):
    for numbers in args:
        print(numbers)
```

```python
print_numbers(1,2,3,4,5,6,7,8,"Henry")
```

```python
### Keywords Arguments
## Hanya untuk key value pairs
def print_details(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(name="Henry", age="28", Country="Indonesia")
```

```python
def print_details(*args,**kwargs):
    for val in args:
        print(f" Positional Argument :{val}")

    for key, value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(1,2,3,4,"Henry", name="Henry", age="28", Country="Indonesia")
```
