### SQL and SQLLite

#### SQL (Structured Query Language) adalah bahasa untuk managing dan manipulating relational databases
#### SQlite itu self-contained, serverless dan zero configuration database engine

```python
import sqlite3
```

```python
## Connect to SQLite Database
# Database akan terbuat di local system tanpa harus instalasi server jika menggunakan sqlite
connection = sqlite3.connect('example.db')
connection
```

```python
cursor = connection.cursor()
```

```python
## Create Table

cursor.execute('''
Create Table If Not Exists employees(
    id Integer Primary Key,
    name Text Not Null,
    age Integer,
    department text
    )
''')

## Commit the changes
connection.commit()
```

```python
cursor.execute('''
Select * from employees
''')
```

```python
## Insert data in sqlite table
cursor.execute('''
Insert Into employees(name,age,department)
values('Henry',22,'Data Scientist')
''')

cursor.execute('''
INSERT INTO employees (name, age, department)
values ('Deri', 25, 'Data Engineer')
''')

cursor.execute('''
INSERT INTO employees (name, age, department)
values ('Dan', 33, 'Finance')
''')

## Commit the changes
connection.commit()
```

```python
## Query data from table
cursor.execute('Select * from employees')
rows=cursor.fetchall()

## Print the query data
for row in rows:
    print(row)
```

```python
## Update data in the table
cursor.execute('''
UPDATE employees
Set age=34
where name="Henry"
''')

connection.commit()
```

```python
## Query data from table
cursor.execute('Select * from employees')
rows=cursor.fetchall()

## Print the query data
for row in rows:
    print(row)
```

```python
## Delete data from the table
cursor.execute('''
Delete from employees
               where name = "Henry"
''')

connection.commit()
```

```python
## Query data from table
cursor.execute('Select * from employees')
rows=cursor.fetchall()

## Print the query data
for row in rows:
    print(row)
```

```python
## Working with Sales Data
# Connect to SQLite database
connection = sqlite3.connect('sales_data.db')
cursor = connection.cursor()

# Create table for sales data
cursor.execute('''
CREATE TABLE IF NOT EXISTS sales(
    id INTEGER PRIMARY KEY,
    date TEXT NOT NULL,
    product TEXT NOT NULL,
    sales INTEGER,
    region TEXT
)
''')

# Insert data into the sales table, cocok untuk 1 data
sales_data = [
    ('2023-01-01', 'Product1', 100, 'North'),
    ('2023-01-02', 'Product2', 200, 'South'),
    ('2023-01-03', 'Product1', 150, 'East'),
    ('2023-01-04', 'Product3', 250, 'West'),
    ('2023-01-05', 'Product2', 300, 'North'),
]

#executemany bisa insert banyak data sekaligus dengan cepat, cocok untuk banyak data
connection.executemany('''
INSERT INTO sales(date, product, sales, region)
                       values(?,?,?,?)
''',sales_data)

connection.commit()
```

```python
## Query data from table
cursor.execute('Select * from sales')
rows=cursor.fetchall()

## Print the query data
for row in rows:
    print(row)
```

```python
## Close the connection
connection.close()
```

```python
## Karena koneksi sudah terclose maka tidak akan bisa di query
## Query data from table
cursor.execute('Select * from sales')
rows=cursor.fetchall()

## Print the query data
for row in rows:
    print(row)
```
