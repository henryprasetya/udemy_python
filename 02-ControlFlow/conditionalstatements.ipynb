```python
## elif
## allow to check multiple condition, it stand for else if

age=19

if age<13:
    print("Kamu anak anak")
elif age<18:
    print("kamu anak muda")
else:
    print("KAmu orang dewasa")
```

```python
## Nested Conditional Statement

num=int(input("enter number"))

if num>0:
    print("The number is positive")
    if num%2==0:
        print ("number is even")
    else:
        print("number is odd")

else:
    print("the number is zero or negative")
```

```python
## Practical Example

# Determine if a year is a leap year using nested conditional statement

year=int(input("Masukkan angka tahun kabisat"))

if year%4==0:
    if year%100==0:
        if year%400==0:
            print(year,"ini tahun kabisat")
        else:
            print(year,"Ini bukan tahun kabisat")
    else:
        print(year,"ini tahun kabisat")
else:
    print(year,"ini tahun kabisat")
```

```python
## Assigment
## Create simple calculator program

angka1=float(input("Masukkan angka pertama"))
operator=input("pilih operasi yang akan digunakan (+, -, *, /)")
angka2=float(input("Masukkan angka kedua"))

if operator == "+":
    hasil = angka1+angka2
elif operator == "-":
    hasil = angka1-angka2
elif operator == "*":
    hasil = angka1*angka2
elif operator == "/":
    if angka2 != 0:
        hasil = angka1/angka2
    else:
        hasil = "Error! Tidak dapat membagi dengan 0"
else:
    hasil = "Operasi invalid"

print("Berikut hasilnya =", hasil)
```
