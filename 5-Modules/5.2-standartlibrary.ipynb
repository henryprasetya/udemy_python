```python
## Data Serialization
# Untuk konversi data misal ke json

import json
data={'name':'Henry','age':25}

json_str=json.dumps(data)        # konversi dari dictionary ke json
print(json_str) 
print(type(json_str))            # lihat tipe data dari json.str

parsed_data=json.loads(json_str) # konversi ke dictionary kembali
print(parsed_data)
print(type(parsed_data))

```

```python
## csv

import csv

with open('example.csv',mode='w',newline='') as file: #membuat file example.csv dengan mode write
    writer=csv.writer(file)                           
    writer.writerow(['name','age'])
    writer.writerow(['Henry',28])

with open('example.csv',mode='r') as file:            #Read seluruh file dari example.csv
    reader=csv.reader(file)
    for row in reader:
        print(row)
```

```python
## datetime

from datetime import datetime,timedelta

now=datetime.now()                  #fungsinya untuk melihat datetime saat ini
print(now)

yesterday=now-timedelta(days=1)     #fungsinya untuk melihat dateime kemarin

print(yesterday)
```

```python
## Regular expression
# Berfungsi untuk mendefinisikan sebuah pola pencarian seperti ctrl+f
# Contoh kasus dibawah ini mencari 

import re

pattern=r'\d+'                      # Cari satu atau lebih digit yang muncul secara berurutan
text='There are 123 apples 456'
match=re.search(pattern,text)
print(match.group())                # Output 123, 456 tidak muncul karena pattern 123 terbaca lebih dahulu
```
